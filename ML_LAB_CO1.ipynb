```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the Iris dataset
df = pd.read_csv("Iris.csv")

# -----------------------------------
# Pandas - Data Handling
# -----------------------------------

print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nShape of Dataset:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())


# -----------------------------------
# NumPy - Numerical Operations
# -----------------------------------

sepal_length = df["SepalLengthCm"].to_numpy()

print("\nSepal Length as NumPy Array:")
print(sepal_length)

print("\nMean Sepal Length:", np.mean(sepal_length))
print("Maximum Sepal Length:", np.max(sepal_length))
print("Minimum Sepal Length:", np.min(sepal_length))


# -----------------------------------
# Matplotlib - Data Visualization
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.plot(df["Id"], df["SepalLengthCm"])

plt.xlabel("Sample ID")
plt.ylabel("Sepal Length (cm)")
plt.title("Sepal Length of Iris Samples")

plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Iris dataset
df = pd.read_csv("Iris.csv")

# Separate sepal width values for each species
setosa = df[df["Species"] == "Iris-setosa"]["SepalWidthCm"]
versicolor = df[df["Species"] == "Iris-versicolor"]["SepalWidthCm"]
virginica = df[df["Species"] == "Iris-virginica"]["SepalWidthCm"]

# Create histograms in a single plot
plt.hist(setosa, bins=10, alpha=0.5, label="Setosa")
plt.hist(versicolor, bins=10, alpha=0.5, label="Versicolor")
plt.hist(virginica, bins=10, alpha=0.5, label="Virginica")

# Add labels and title
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Frequency")
plt.title("Sepal Width Distribution by Iris Species")

# Display legend
plt.legend()

# Display plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Iris dataset
df = pd.read_csv("Iris.csv")

# Separate sepal length for each species
setosa = df[df["Species"] == "Iris-setosa"]["SepalLengthCm"]
versicolor = df[df["Species"] == "Iris-versicolor"]["SepalLengthCm"]
virginica = df[df["Species"] == "Iris-virginica"]["SepalLengthCm"]

# Create boxplot
plt.boxplot(
    [setosa, versicolor, virginica],
    labels=["Setosa", "Versicolor", "Virginica"]
)

# Add labels and title
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.title("Sepal Length Comparison by Iris Species")

# Display plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Iris dataset
df = pd.read_csv("Iris.csv")

# Separate data for each species
setosa = df[df["Species"] == "Iris-setosa"]
versicolor = df[df["Species"] == "Iris-versicolor"]
virginica = df[df["Species"] == "Iris-virginica"]

# Create scatterplots
plt.scatter(
    setosa["PetalLengthCm"],
    setosa["PetalWidthCm"],
    label="Setosa"
)

plt.scatter(
    versicolor["PetalLengthCm"],
    versicolor["PetalWidthCm"],
    label="Versicolor"
)

plt.scatter(
    virginica["PetalLengthCm"],
    virginica["PetalWidthCm"],
    label="Virginica"
)

# Add labels and title
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Petal Length vs Petal Width by Iris Species")

# Display legend
plt.legend()

# Display plot
plt.show()
```
